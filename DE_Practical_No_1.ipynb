```python
import csv
import json
import xml.etree.ElementTree as ET
from bs4 import BeautifulSoup


# Create a text file
with open("students.txt", "w") as file:
    file.write("101, Rahul, 85\n")
    file.write("102, Priya, 92\n")
    file.write("103, Amit,\n")
    file.write("104, Sneha, 78\n")
    file.write("105, Rohan, 150\n")


# Create a CSV file
with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["Roll No", "Name", "Marks"])
    writer.writerow([101, "Rahul", 85])
    writer.writerow([102, "Priya", 92])
    writer.writerow([103, "Amit", ""])
    writer.writerow([104, "Sneha", 78])
    writer.writerow([105, "Rohan", 150])


# Create an HTML file
html_data = """
<html>
<body>

<h2>Student Details</h2>

<table>
<tr>
<th>Roll No</th>
<th>Name</th>
<th>Marks</th>
</tr>

<tr>
<td>101</td>
<td>Rahul</td>
<td>85</td>
</tr>

<tr>
<td>102</td>
<td>Priya</td>
<td>92</td>
</tr>

<tr>
<td>103</td>
<td>Amit</td>
<td></td>
</tr>

<tr>
<td>104</td>
<td>Sneha</td>
<td>78</td>
</tr>

</table>

</body>
</html>
"""

with open("students.html", "w") as file:
    file.write(html_data)


# Create an XML file
root = ET.Element("students")

student = ET.SubElement(root, "student")
ET.SubElement(student, "roll_no").text = "101"
ET.SubElement(student, "name").text = "Rahul"
ET.SubElement(student, "marks").text = "85"

student = ET.SubElement(root, "student")
ET.SubElement(student, "roll_no").text = "102"
ET.SubElement(student, "name").text = "Priya"
ET.SubElement(student, "marks").text = "92"

student = ET.SubElement(root, "student")
ET.SubElement(student, "roll_no").text = "103"
ET.SubElement(student, "name").text = "Amit"
ET.SubElement(student, "marks").text = ""

tree = ET.ElementTree(root)
tree.write("students.xml")


# Create a JSON file
students = [
    {
        "roll_no": 101,
        "name": "Rahul",
        "marks": 85
    },
    {
        "roll_no": 102,
        "name": "Priya",
        "marks": 92
    },
    {
        "roll_no": 103,
        "name": "Amit",
        "marks": None
    },
    {
        "roll_no": 104,
        "name": "Sneha",
        "marks": 78
    }
]

with open("students.json", "w") as file:
    json.dump(students, file, indent=4)


# --------------------------------------------------
# 1. Parse Text File
# --------------------------------------------------

print("\n1. TEXT FILE")
print("--------------------")

with open("students.txt", "r") as file:

    for line in file:

        line = line.strip()

        if line == "":
            continue

        data = line.split(",")

        roll_no = data[0].strip()
        name = data[1].strip()
        marks = data[2].strip()

        if marks == "":
            print("Missing marks for:", name)

        else:
            marks = int(marks)

            if marks < 0 or marks > 100:
                print("Anomaly found for:", name)

            else:
                print("Roll No:", roll_no)
                print("Name:", name)
                print("Marks:", marks)
                print()


# --------------------------------------------------
# 2. Parse CSV File
# --------------------------------------------------

print("\n2. CSV FILE")
print("--------------------")

with open("students.csv", "r") as file:

    reader = csv.DictReader(file)

    for row in reader:

        name = row["Name"]
        marks = row["Marks"]

        if marks == "":
            print("Missing marks for:", name)

        else:
            marks = int(marks)

            if marks < 0 or marks > 100:
                print("Anomaly found for:", name)

            else:
                print("Name:", name)
                print("Marks:", marks)


# --------------------------------------------------
# 3. Parse HTML File
# --------------------------------------------------

print("\n3. HTML FILE")
print("--------------------")

with open("students.html", "r") as file:

    html_data = file.read()

soup = BeautifulSoup(html_data, "html.parser")

table = soup.find("table")

rows = table.find_all("tr")

for row in rows[1:]:

    columns = row.find_all("td")

    roll_no = columns[0].text.strip()
    name = columns[1].text.strip()
    marks = columns[2].text.strip()

    if marks == "":
        print("Missing marks for:", name)

    else:
        print("Roll No:", roll_no)
        print("Name:", name)
        print("Marks:", marks)


# --------------------------------------------------
# 4. Parse XML File
# --------------------------------------------------

print("\n4. XML FILE")
print("--------------------")

tree = ET.parse("students.xml")

root = tree.getroot()

for student in root.findall("student"):

    roll_no = student.find("roll_no").text
    name = student.find("name").text
    marks_element = student.find("marks")

    if marks_element.text is None or marks_element.text.strip() == "":
        print("Missing marks for:", name)

    else:
        marks = int(marks_element.text)

        print("Roll No:", roll_no)
        print("Name:", name)
        print("Marks:", marks)


# --------------------------------------------------
# 5. Parse JSON File
# --------------------------------------------------

print("\n5. JSON FILE")
print("--------------------")

with open("students.json", "r") as file:

    data = json.load(file)

for student in data:

    roll_no = student["roll_no"]
    name = student["name"]
    marks = student["marks"]

    if marks is None:
        print("Missing marks for:", name)

    else:
        print("Roll No:", roll_no)
        print("Name:", name)
        print("Marks:", marks)


print("\nAll files were parsed successfully.")
```
