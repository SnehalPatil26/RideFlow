```python
import csv
from pyspark.sql import SparkSession
from pyspark.sql.functions import avg, sum, count

# Create sample sales.csv file
with open('sales.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(["Customer_ID", "Category", "Sales"])
    writer.writerow([101, "Electronics", 6000])
    writer.writerow([102, "Furniture", 3000])
    writer.writerow([103, "Electronics", 1500])
    writer.writerow([101, "Electronics", 6000]) # Duplicate row to test dropDuplicates
    writer.writerow([104, "Office Supplies", 800])

# Create sample customers.csv file
with open('customers.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(["Customer_ID", "Customer_Name"])
    writer.writerow([101, "Alice"])
    writer.writerow([102, "Bob"])
    writer.writerow([103, "Charlie"])
    writer.writerow([104, "David"])

# Create Spark Session
spark = SparkSession.builder \
    .appName("PySpark Practical 8") \
    .getOrCreate()

# ---------------------------------------------------
# 1. Read CSV Dataset
# ---------------------------------------------------

df = spark.read.csv(
    "sales.csv",
    header=True,
    inferSchema=True
)

print("Original Dataset:")
df.show()

# ---------------------------------------------------
# 2. Display Schema
# ---------------------------------------------------

print("Dataset Schema:")
df.printSchema()

# ---------------------------------------------------
# 3. Filtering
# ---------------------------------------------------

print("Filtered Records:")
df.filter(df.Sales > 5000).show()

# ---------------------------------------------------
# 4. Grouping and Aggregation
# ---------------------------------------------------

print("Total Sales by Category:")
df.groupBy("Category") \
  .agg(sum("Sales").alias("Total_Sales")) \
  .show()

# ---------------------------------------------------
# 5. Remove Duplicate Records
# ---------------------------------------------------

df_unique = df.dropDuplicates()

print("Dataset After Removing Duplicates:")
df_unique.show()

# ---------------------------------------------------
# 6. Read Second Dataset
# ---------------------------------------------------

customer_df = spark.read.csv(
    "customers.csv",
    header=True,
    inferSchema=True
)

print("Customer Dataset:")
customer_df.show()

# ---------------------------------------------------
# 7. Join Two Datasets
# ---------------------------------------------------

joined_df = df.join(
    customer_df,
    df.Customer_ID == customer_df.Customer_ID,
    "inner"
)

print("Joined Dataset:")
joined_df.show()

# ---------------------------------------------------
# 8. Average Sales by Product Category
# ---------------------------------------------------

print("Average Sales by Product Category:")

df.groupBy("Category") \
  .agg(avg("Sales").alias("Average_Sales")) \
  .show()

# Stop Spark
spark.stop()
```

Result
The PySpark program successfully:
1. Read the CSV dataset.
2. Displayed the dataset schema.
3. Performed filtering operations.
4. Performed grouping and aggregation.
5. Removed duplicate records.
6. Joined two datasets using Customer_ID.
7. Calculated average sales for each product category.
